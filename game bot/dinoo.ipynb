```python
import pyautogui
import time
import ctypes

# =========================
# 1. Windows 螢幕取色設定
# =========================

user32 = ctypes.windll.user32
gdi32 = ctypes.windll.gdi32

# 取得螢幕 DC
hdc = user32.GetDC(0)

# 設定 GetPixel 的參數與回傳值
gdi32.GetPixel.argtypes = [
    ctypes.c_void_p,
    ctypes.c_int,
    ctypes.c_int
]
gdi32.GetPixel.restype = ctypes.c_uint32


# =========================
# 2. 取得指定座標的顏色
# =========================

def get_pixel(x, y):
    return gdi32.GetPixel(hdc, x, y)


# =========================
# 3. 恐龍與仙人掌座標
# =========================

# 恐龍中心位置（你測量的座標）
恐龍_X = 43
恐龍_Y = 328

# 矮仙人掌高度範圍
仙人掌頂部_Y = 297
仙人掌底部_Y = 342

# 偵測仙人掌的垂直位置
偵測_Y = 320

# 偵測仙人掌的水平位置
# 仙人掌碰到恐龍時，左側約 X=58
# 因此先提早在 X=100 偵測
偵測_X = 100


# =========================
# 4. 開始遊戲前的準備
# =========================

print("請先把 Chrome 恐龍遊戲視窗放好。")
input("確認遊戲畫面準備好後，按 Enter 開始：")

print("3 秒後開始偵測，請準備好！")
time.sleep(3)

# 記錄沒有仙人掌時的背景顏色
背景顏色 = get_pixel(偵測_X, 偵測_Y)

print("背景顏色：", hex(背景顏色))
print("開始遊戲！")

# 開始遊戲
pyautogui.press("space")

time.sleep(0.5)


# =========================
# 5. 自動跳躍主程式
# =========================

障礙物已觸發 = False

try:
    while True:

        # 取得偵測位置目前的顏色
        目前顏色 = get_pixel(偵測_X, 偵測_Y)

        # 如果取色失敗，就略過這次判斷
        if 目前顏色 == 0xFFFFFFFF:
            time.sleep(0.01)
            continue

        # 偵測到仙人掌
        if 目前顏色 != 背景顏色 and not 障礙物已觸發:

            print("偵測到障礙物，跳躍！")

            # 按下空白鍵
            pyautogui.keyDown("space")
            time.sleep(0.08)
            pyautogui.keyUp("space")

            # 避免同一個仙人掌重複觸發
            障礙物已觸發 = True

        # 仙人掌離開偵測位置
        # 重新開啟下一次偵測
        elif 目前顏色 == 背景顏色:

            障礙物已觸發 = False

        # 縮短等待時間，提高偵測頻率
        time.sleep(0.01)

except KeyboardInterrupt:
    print("程式已停止。")

finally:
    # 釋放 Windows 螢幕 DC
    user32.ReleaseDC(0, hdc)
```
