# 簡介
串列是一個可以儲存多個元素(變數,浮點數,字串)的容器
串列中每個元素都是用逗點隔開,左右有[]

```python
scores=[78,29,56,80,99]
print(scores)
```

# 索引值
可以用索引值(index)來存取列中的元素
索引值由0開始

```python
print(scores[0])#印出第一個元素
print(scores[1])#印出第二個元素
print(scores[2])
```

# 串列內容的個數
list是length(長度)的意思

```python
print(len(scores))
print(scores[4])#取得最後一個元素
print(scores[5-1])#取得最後一個元素
print(scores[-1])#取得最後一個元素
print(scores[len(scores)-1])
```

# 新增元素
使用append()新增元素到串列的最尾端

```python
print(scores)
scores[2]=70 #將70指定給索引值為2的元素
print(scores)
```

```python
weights=[12]*6
print(weights)
```
