Python:
- 懶惰是種美德
- 由0開頭
- 包含頭，不包含尾

```python
print(list(range(5)))
#區間由0開始到5-1結束
print(list(range(0,5)))
#區間由1開始到5-1結束
print(list(range(1,5)))
#區間由1開始到5-1結束，每次跳2個
print(list(range(1,5,2)))
```
